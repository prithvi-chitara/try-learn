```python
# 2 ways to traverse a list
```

```python
# type 1 itemwise
# regular list
l = [1,2,3,4]
for i in l:
  print(i)
```

```python
# type 2 index wise
l = [1,2,3,4]
for i in range(0,len(l)):
  print("index",i,"list value is",end=" ")
  print(l[i])
```
