```python
# Q4. type 1 .  check given string is palindrome or not
```

```python
s = input("enter string")
if s == s[::-1]:
  print("palindrome ",s)
else:
  print("not palindrome",s)
```

```python
# type 2
s = input("enter string")
flag = True
for i in range(0,len(s)//2):
  if s[i]!= s[len(s)-i-1]:
    flag = False
    print("not palindrome")
    break

if flag:
  print("palindrome")
```

```python
num = int(input("enter a number "))
resu = ""
digit = "0123456789"
while num!=0:
  resu = digit[num%10]+resu
  num = num//10
print(resu)
print(type(resu))

```
